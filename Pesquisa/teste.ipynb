```python
def ler_xls(arquivo_excel):
    workbook = xlrd.open_workbook(arquivo_excel)
    # Escolher a primeira planilha (índice 0)
    sheet = workbook.sheet_by_index(0)

    # Ler os dados da planilha
    dados = [sheet.row_values(i) for i in range(sheet.nrows)]

    # Converter para DataFrame
    return pd.DataFrame(dados[1:], columns=dados[0])

def salvar_xls(dataframe):
    # Especificar o nome do arquivo CSV de saída
    arquivo_csv = 'saida.xlsx'

    # Salvar o DataFrame como um arquivo CSV
    dataframe.to_excel(arquivo_csv, index= False)
    #new_df.to_csv(arquivo_csv, index=False)
```
