```python
import pyautogui
import time

def collect_target(img):
    x, y = pyautogui.locateCenterOnScreen('sample/'+f'{img}.png', confidence= 0.5)
    pyautogui.click(x,y)

def wait_loading_page(img, wait_time = 40):
    time_pass = 0
    while time_pass < wait_time:
        try:
            x, y = pyautogui.locateCenterOnScreen('sample/'+f'{img}.png', confidence= 0.6)
        except:
            time.sleep(1)
            pass
        else:
            time.sleep(2)
            return

def move(dir, length):
    pyautogui.keyDown(dir)
    time.sleep(length)
    pyautogui.keyUp(dir)

def foward_movement (directions):
    for dir in directions:
        move(dir['key_name'], dir['time_press'])

def skip_land(proportion = 1):
    '''
    The  proportion parameter its 1 for 100% time.
    
    The base time is 3 seconds
    '''
    SKIP_LAND_TIME_WALK = 3
    pyautogui.keyDown('right')
    time.sleep(SKIP_LAND_TIME_WALK*proportion)
    pyautogui.keyUp('right')

    wait_loading_page("ancora")

def completed_movement(directions): #The name should be changed for 'harvest'
    foward_movement(directions)
    
    time.sleep(0.5)
    collect_target("galinheiro")
    backward_movement(directions)

    time.sleep(0.5)
    skip_land(0.6)
    


def backward_movement(directions):
    for dir in reversed(directions[1:]):
        if dir['key_name'] == 'up':
            key_name = 'down'
        elif dir['key_name'] == 'down':
            key_name = 'up'
        elif dir['key_name'] == 'left':
            key_name = 'right'
        elif dir['key_name'] == 'right':
            key_name = 'left'
        move(key_name, dir['time_press'])
 
```

```python
import json
from pathlib import Path

json_file_path = Path('rotas_galinheiro.json')
json_data = []
if json_file_path.exists():
    with open(json_file_path, 'r') as existing_json_file:
        json_data = json.load(existing_json_file)

INITIAL_LAND_NUMBER = 1077
time.sleep(5)
for actual_land in range(INITIAL_LAND_NUMBER, INITIAL_LAND_NUMBER+10):
    print(f"Land Atual: {actual_land}")
    posicao = next((index for index, item in enumerate(json_data) if int(item["land_number"]) == actual_land), -1)
    if posicao != -1:
        print("\t Indo colher os ovos")
        completed_movement(json_data[posicao]['directions'])
    else:
        skip_land()
```
