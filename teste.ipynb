```python
import pyautogui
import time


def add_chickens(confidence = 0.5, num_chickens = 2):
    if pyautogui.locateCenterOnScreen('sample/'+f'galinheiro.png', confidence= 0.5):
        pyautogui.hotkey('6')
        x, y = pyautogui.locateCenterOnScreen('sample/'+f'galinheiro.png', confidence= 0.5)
        for _ in range(0,num_chickens):
            pyautogui.click(x,y)
            time.sleep(1.5)
        pyautogui.hotkey('6')


def collect_target(img):
    pyautogui.move(0,0, duration=0.5)
    x, y = pyautogui.locateCenterOnScreen('sample/'+f'{img}.png', confidence= 0.5)
    pyautogui.click(x,y)

def wait_loading_page(img, wait_time = 40):
    time_pass = 0
    while time_pass < wait_time:
        try:
            x, y = pyautogui.locateCenterOnScreen('sample/'+f'{img}.png', confidence= 0.6)
        except:
            time.sleep(1)
            pass
        else:
            time.sleep(1)
            return

def move(dir, length):
    pyautogui.keyDown(dir)
    time.sleep(length)
    pyautogui.keyUp(dir)

def foward_movement (directions):
    wait_loading_page("ancora")
    for dir in directions:
        move(dir['key_name'], dir['time_press'])

def skip_land(proportion = 1):
    '''
    The  proportion parameter its 1 for 100% time.
    
    The base time is 3 seconds
    '''
    SKIP_LAND_TIME_WALK = 3
    pyautogui.keyDown('right')
    time.sleep(SKIP_LAND_TIME_WALK*proportion)
    pyautogui.keyUp('right')

    wait_loading_page("ancora")

def completed_movement(directions, first_land = False, last_land = False): #The name should be changed for 'harvest'
    foward_movement(directions)
    
    time.sleep(0.1)
    add_chickens(num_chickens=1)
    collect_target("galinheiro")
    backward_movement(directions, first_land)

    time.sleep(0.1)
    if not last_land:
        skip_land(1)
    
    
def backward_movement(directions, first_land = False):
    aux = 1
    if first_land:
        aux = 0
    for dir in reversed(directions[aux:]): #Retirei o direction[1:]
        if dir['key_name'] == 'up':
            key_name = 'down'
        elif dir['key_name'] == 'down':
            key_name = 'up'
        elif dir['key_name'] == 'left':
            key_name = 'right'
        elif dir['key_name'] == 'right':
            key_name = 'left'
        move(key_name, dir['time_press'])
 

def walk_with_mouse(img, time_press, factor = 60):
    verify = False
    for _ in range(0, time_press):
        try:
            x, y = pyautogui.locateCenterOnScreen('sample/'+f'{img}.png', confidence= 0.6)
            pyautogui.mouseDown(x,y+factor)
            time.sleep(1)
            pyautogui.mouseUp()
        except:
            pass
        else:
            verify = True
    return verify

def go_through_door():
    for _ in range (0,3):
      try:
        x, y = pyautogui.locateCenterOnScreen('sample/'+f'office_door.png', confidence= 0.6)
        pyautogui.mouseDown(x,y)
        time.sleep(3)
        pyautogui.mouseUp()
      except:
          pass
      else:
          return True
      time.sleep(2)
    
def go_to_land_office():
    
    with open("basics_movements.json", 'r') as existing_json_file:
        BASIC_MOVEMENTS = json.load(existing_json_file)

    if walk_with_mouse('statue_1', 4):
        foward_movement(BASIC_MOVEMENTS['statue_up'])
        go_through_door()

    elif walk_with_mouse('statue_2', 4):
        foward_movement(BASIC_MOVEMENTS['statue_down'])
        go_through_door()

def set_land_number_infiniportal(land_number):
    x, y = pyautogui.locateCenterOnScreen('sample/'+f'infiniportal.png', confidence= 0.6)
    pyautogui.click(x,y)
    time.sleep(0.5)
    pyautogui.write(str(land_number))


def go_to_terravilla():
    x, y = pyautogui.locateCenterOnScreen('sample/'+'terravilla_icon.png', confidence= 0.6)
    pyautogui.click(x,y)
    time.sleep(0.5)
    x, y = pyautogui.locateCenterOnScreen('sample/'+'go_to_terravila.png', confidence= 0.8)
    pyautogui.click(x,y)

def travel_to_specific_land(land_number):
    with open("basics_movements.json", 'r') as existing_json_file:
        BASIC_MOVEMENTS = json.load(existing_json_file)
    go_to_terravilla()
    wait_loading_page('ancora')
    go_to_land_office()
    foward_movement(BASIC_MOVEMENTS['go_infiniportal'])
    set_land_number_infiniportal(land_number)
    time.sleep(0.5)
    pyautogui.hotkey('enter')
```

```python
import json

with open("rotas_galinheiro.json", 'r') as existing_json_file:
    LAND_MOVEMENTS = json.load(existing_json_file)

with open("rotations.json", 'r') as existing_json_file:
    ROTATIONS = json.load(existing_json_file)

vector_rotations = list(ROTATIONS.values())
time.sleep(4)
for rotation in vector_rotations:
    travel_to_specific_land(rotation['initial_land'])
    time.sleep(1)
    for actual_land in range(rotation['initial_land'], rotation['final_land']+1):
        print(f"Land Atual: {actual_land}")
        posicao = next((index for index, item in enumerate(LAND_MOVEMENTS) if int(item["land_number"]) == actual_land), -1)

        if posicao != -1 and actual_land != rotation['final_land']:
            print("\t Indo colher os ovos")
            if actual_land == rotation['initial_land']:
                completed_movement(LAND_MOVEMENTS[posicao]['directions'], True)
            else:
                completed_movement(LAND_MOVEMENTS[posicao]['directions'])
        elif posicao != -1 and actual_land == rotation['final_land']:
            print("\t Indo colher os ovos")
            completed_movement(LAND_MOVEMENTS[posicao]['directions'], last_land= True)
        else:
            skip_land()
```
